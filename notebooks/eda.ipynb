# Data Preparation

```python
# Convert the date column to datetime objects
df["date"] = pd.to_datetime(df["date"])
```

```python
# Filter rows where the date is between January 1, 2018, and December 31, 2026
filtered_df = df[(df["date"] >= "2026-01-01") & (df["date"] <= "2026-12-31")]

# View the result
print(filtered_df.shape)

filtered_df[filtered_df["admin2"] == "Taunggyi"].T
```

```python
rice = df[
    (df["commodity"] == "Rice (low quality)") &
    (df["pricetype"] == "Retail")
].copy()

rice["date"] = pd.to_datetime(rice["date"])

periods = {
    "2008-2026": rice,
    "2015-2026": rice[rice["date"] >= "2015-01-01"],
    "2020-2026": rice[rice["date"] >= "2020-01-01"]
}

for name, data in periods.items():
    print(f"\n{name}")
    print("Rows:", len(data))
    print("Markets:", data["market"].nunique())
    print("Regions:", data["admin1"].nunique())
    print("Date range:", data["date"].min(), data["date"].max())
```
